```python
import os
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

try:
    # Colab Secrets سے کی حاصل کرنے کی کوشش کریں
    api_key = userdata.get('GEMINI_API_KEY')
    if api_key:
        os.environ["GOOGLE_API_KEY"] = api_key.strip()
    else:
        os.environ["GOOGLE_API_KEY"] = "YOUR_GEMINI_API_KEY_HERE"
except Exception:
    os.environ["GOOGLE_API_KEY"] = "YOUR_GEMINI_API_KEY_HERE".strip()

# چیک کریں کہ آیا کی واقعی سیٹ ہوئی ہے اور اس کا فارمیٹ ٹھیک ہے
current_key = os.environ.get("GOOGLE_API_KEY", "")
print(f"کی کی لمبائی (Length): {len(current_key)}")
if not current_key.startswith("AIzaSy"):
    print("وارننگ: عام طور پر گوگل کیز 'AIzaSy' سے شروع ہوتی ہیں۔ آپ کی کی مختلف فارمیٹ میں ہے۔")

# 1. ماڈل ڈیفائن کریں (ہم نے ماڈل کا نام 'gemini-1.5-flash' برقرار رکھا ہے)
model = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

# 2. پرامپٹ ٹیمپلیٹ بنائیں
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "آپ ایک ماہر سوشل میڈیا مارکیٹر ہیں۔"),
    ("user", "مجھے '{topic}' کے بارے میں ایک دلچسپ فیس بک پوسٹ لکھ کر دو۔")
])

# 3. آؤٹ پٹ پارسر
output_parser = StrOutputParser()

# 4. پائپ لائن بنانا
chain = prompt_template | model | output_parser

# 5. چین کو چلانا اور کوٹہ چیک کرنا
try:
    result = chain.invoke({"topic": "ارٹیفیشل انٹیلی جنس اور فیوچر جابز"})
    print("--- Gemini (Free) آؤٹ پٹ ---")
    print(result)
except Exception as e:
    print(f"ایرر کی تفصیل: {e}")
    if "API_KEY_INVALID" in str(e) or "not found" in str(e).lower():
        print("\n[تجویز] براہ کرم https://aistudio.google.com/ پر جائیں اور ایک نئی اور درست API Key بنا کر اسے استعمال کریں۔")
```
